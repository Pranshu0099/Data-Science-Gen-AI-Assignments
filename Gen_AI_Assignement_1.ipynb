**Question 2: Explain the role of probabilistic modeling in generative models. How do these models differ from discriminative models?**

**Answer:**

1. The Role of Probabilistic Modeling in Generative ModelsAt its core, Generative AI is an exercise in probability distribution estimation. Real-world data—whether it is human language, medical scans, or photographic images—is not random noise; it follows an underlying, highly structured, but unknown true data distribution, denoted as:$$p_{\text{data}}(\mathbf{x})$$The goal of a generative model is to construct a parameterized probability model $p_\theta(\mathbf{x})$ that approximates $p_{\text{data}}(\mathbf{x})$ as closely as possible, allowing the system to sample novel data points that look authentic.Probabilistic modeling plays several fundamental roles:

A. Density Estimation (Learning What is "Likely"): The model assigns high probability density to realistic instances and near-zero probability to nonsensical noise.

B. B. Sampling and Stochastic Generation: Instead of generating deterministic, repetitive answers, generative models use probability distributions to sample diverse outputs.

C. Latent Space Representation & Manifold Learning: Complex high-dimensional data (e.g., $1024 \times 1024$ image pixels) actually resides on a much lower-dimensional manifold.

**2. How Generative Models Differ from Discriminative Models:**

1. Core Objective and Learned Distribution

Discriminative models learn the conditional probability distribution $P(Y \mid X)$, mapping high-dimensional input features $X$ to a predefined target label $Y$. Their explicit goal is classification or regression. In contrast, generative models estimate either the joint distribution $P(X, Y)$ or the marginal distribution $P(X)$, learning how the features themselves are constructed.

2. Decision Boundaries vs. Data Distribution

A discriminative model only discovers the boundary or hyperplane separating classes. It does not need to learn what makes an object look complete or authentic; it only needs enough distinguishing features to separate a dog from a cat. A generative model captures the full data distribution of every class, allowing it to synthesize a completely new example from scratch.

3. Input-to-Output Flow

Discriminative models compress rich, high-dimensional inputs into low-dimensional outputs, such as categorical labels, bounding box coordinates, or risk scores. Generative models operate in reverse or across complex spaces: they take low-dimensional inputs (such as random noise vectors or short text prompts) and expand them into high-dimensional data like full-length essays, source code, or high-resolution images.

**Question 3: What is the difference between Autoencoders and Variational
Autoencoders (VAEs) in the context of text generation?**

**Answer:**

1. Deterministic Mapping vs. Probabilistic Distribution

Standard Autoencoder (AE):Operates as a deterministic compression system. The encoder network takes a sequence of text tokens and compresses it into a single, fixed point vector $\mathbf{z}$ in latent space:$$\mathbf{z} = \text{Encoder}(\mathbf{x})$$


Variational Autoencoder (VAE):Operates as a probabilistic model. Instead of mapping a sentence to a single point, the encoder predicts the parameters of a probability distribution—typically the mean vector $\boldsymbol{\mu}$ and log-variance vector $\log(\boldsymbol{\sigma}^2)$:$$q_\phi(\mathbf{z} \mid \mathbf{x}) = \mathcal{N}(\boldsymbol{\mu}, \boldsymbol{\sigma}^2)$$

2. Latent Space Geometry and Continuity

Standard Autoencoder: Because there is no constraint enforcing where representations are stored, the latent space develops wide gaps, disjointed clusters, and empty regions ("holes").

While the decoder can reconstruct the specific training sentences it has seen, picking an arbitrary point in the latent space or interpolating between two sentences produces nonsensical, ungrammatical token sequences.

Variational Autoencoder: Introduces a regularization term via Kullback–Leibler (KL) Divergence in the loss function:$$\mathcal{L}_{\text{VAE}} = \mathcal{L}_{\text{reconstruction}} + D_{\text{KL}}(q_\phi(\mathbf{z} \mid \mathbf{x}) \parallel p(\mathbf{z}))$$(where $p(\mathbf{z})$ is typically a standard normal distribution $\mathcal{N}(0, \mathbf{I})$).

3. Generative Capability for Novel Sentences

Standard Autoencoders Cannot Act as True Generative Models: An AE has no defined prior distribution $p(\mathbf{z})$. There is no mathematical way to know where valid text representations live in the latent space, making random sentence synthesis impossible. It is strictly used for dimensionality reduction, feature extraction, or text hashing.

VAEs Enable Controlled Text Synthesis and Semantic Interpolation: To generate a new sentence, one simply draws a random sample from the standard normal prior $\mathbf{z} \sim \mathcal{N}(0, \mathbf{I})$ and passes it through the text decoder.VAEs also enable smooth semantic interpolation: traversing a straight line in latent space from the embedding of "The weather is cold today" to "The sun is shining brightly" produces grammatically coherent intermediate sentences with gradual semantic transitions.

4. Unique Text-Specific Challenge: Posterior Collapse (KL Collapse)

When adapting VAEs to natural language processing (using autoregressive decoders like RNNs or Transformers), VAEs face a distinct failure mode known as posterior collapse.

Because autoregressive text decoders are powerful enough to predict the next word based purely on previous generated tokens, the model can minimize its total loss by driving the KL divergence to zero ($D_{\text{KL}} \to 0$) and ignoring the latent code $\mathbf{z}$ entirely.

When collapse occurs, the VAE degenerates into a standard unconditional language model, losing its ability to use latent representations for controlled text generation.

Standard Autoencoders do not suffer from KL collapse because they have no KL penalty and rely entirely on the latent vector $\mathbf{z}$ for reconstruction.

**Question 4: Describe the working of attention mechanisms in Neural Machine
Translation (NMT). Why are they critical?**

**Answer:**

1. How the Attention Mechanism Works in NMT

In traditional sequence-to-sequence (Seq2Seq) translation models, an encoder processes a source sentence and compresses it into a single, fixed-size vector. The attention mechanism removes this bottleneck by allowing the decoder to selectively look back at all intermediate encoder states during each translation step.

The working mechanism follows four mathematical steps at every decoding step $t$:

A. Computing Alignment Scores: For the current target word being generated, the decoder hidden state $\mathbf{s}_t$ is compared against every source token's encoder hidden state $\mathbf{h}_i$ ($i = 1, \dots, T_x$) to determine how relevant each source word is.

B. Calculating Attention Weights (Softmax Normalization): The raw scalar alignment scores $e_{t,i}$ are passed through a Softmax function across all source positions. This transforms them into a valid probability distribution where each weight represents the percentage of focus to allocate to that word:$$\alpha_{t,i} = \frac{\exp(e_{t,i})}{\sum_{j=1}^{T_x} \exp(e_{t,j})}, \quad \text{where } \sum_{i=1}^{T_x} \alpha_{t,i} = 1.0$$

C. Generating the Dynamic Context VectorThe model computes a weighted sum of the encoder's source representations (or Values $\mathbf{V}$ in Transformers) using the normalized attention weights $\alpha_{t,i}$:$$\mathbf{c}_t = \sum_{i=1}^{T_x} \alpha_{t,i} \mathbf{h}_i$$This produces a tailored, dynamic context vector $\mathbf{c}_t$ specifically focused on the source words most relevant to the current translation step.

D. Predicting the Next Target WordThe dynamic context vector $\mathbf{c}_t$ is concatenated with the current decoder state $\mathbf{s}_t$. The combined vector is projected through a linear layer followed by a Softmax over the target vocabulary to predict the next token:$$\hat{y}_t = \text{softmax}\left(\mathbf{W}_o [\mathbf{s}_t; \mathbf{c}_t] + \mathbf{b}_o\right)$$

**2. Why Attention Mechanisms Are Critical**  

Eliminating the Information Bottleneck: Without attention, an encoder must compress a sentence of 10, 40, or 80 words into a single vector of fixed dimension (e.g., 512 numbers).

Mitigating the Vanishing Gradient Problem: In standard recurrent architectures (RNN/LSTM), gradients must backpropagate through dozens of recurrent steps across the sentence, causing early words to be forgotten.

Handling Non-Sequential Word Alignment & Reordering: Natural languages do not follow the same word order. For example, English adjective-noun order ("the white house") reverses in French or Spanish ("la maison blanche" / "la casa blanca").

Interpretability and Visual Alignments: The attention weight matrix $\alpha$ provides a transparent, interpretable heat map of how the model correlates source tokens with target tokens, making translation errors easy to diagnose and audit.

**Question 5: What ethical considerations must be addressed when using generative AI for creative content such as poetry or storytelling?**

**Answer:**

1. Copyright, Intellectual Property, and Fair Use

Training Data Provenance: Generative AI models are trained on billions of scraped literary works, poems, novels, and scripts without the explicit consent, opt-out mechanisms, or remuneration of the original authors.

Style Mimicry and Derivative Works: Using prompts that emulate living creators directly (e.g., "Write a poem in the distinct voice of Maya Angelou" or "Draft a chapter in the prose style of Brandon Sanderson") risks diluting their unique artistic identity, lowering the commercial value of their human labor, and presenting unsettled legal questions around copyright infringement.

2. Plagiarism, Authorship, and Attribution Transparency

Verbatim or Near-Verbatim Memorization: Large Language Models can inadvertently reproduce memorized fragments of pre-existing, copyrighted prose or poetry without citing the original author.

Transparency and Disclosure: Presenting AI-synthesized fiction, scripts, or poems as purely human-authored work deceives readers, undermines creative integrity, and subverts literary competitions or academic standards. Clearly labeling AI involvement (e.g., co-written, assisted, or synthesized) is critical.

3. Algorithmic Bias and Cultural Erasure

Stereotyping and Archetype Reinforcement: Generative models reproduce historical tropes and biases present in web-scale literature—such as perpetuating gendered portrayals in romance, vilifying marginalized identities, or defaulting characters to Western, Eurocentric cultural contexts.

Homogenization of Voice: Reliance on standard generative models tends to smooth out regional idioms, minority dialects (e.g., AAVE), and non-standard narrative structures, creating an art form centered on generalized, corporate-safe linguistic conventions.

4. Economic Impact and the Livelihood of Human Artists

Devaluation of Creative Labor: Mass-producing low-cost or free synthetic prose and poetry risks displacing freelance writers, commercial copywriters, script readers, and emerging poets.

Monopolization of Market Channels: Unregulated automated publishing has flooded self-publishing storefronts (e.g., Amazon KDP) and literary magazines with low-effort synthetic submissions, crowding out emerging human writers and overwhelming human editorial staff.

5. Generation of Harmful, Toxic, or Exploitative Content

Hate Speech and Violence: Creative storytelling mechanisms can be manipulated (via "jailbreaks" or fictional roleplay framing) to generate graphic violence, non-consensual explicit stories, hate speech, or harassment directed at real individuals.

Defamation in Narrative Contexts: Blending real living people into fictionalized biographical stories or docu-novels can lead to factual hallucinations that damage reputations or spread malicious misinformation.

```python
'''
Question 6: Use the following small text dataset to train a simple Variational
Autoencoder (VAE) for text reconstruction:
["The sky is blue", "The sun is bright", "The grass is green",
"The night is dark", "The stars are shining"]
1. Preprocess the data (tokenize and pad the sequences).
2. Build a basic VAE model for text reconstruction.
3. Train the model and show how it reconstructs or generates similar sentences.
Include your code, explanation, and sample outputs.
'''

import torch
import torch.nn as nn
import torch.optim as optim

# Set manual seed for reproducibility
torch.manual_seed(42)

# Raw Dataset
corpus = [
    "The sky is blue",
    "The sun is bright",
    "The grass is green",
    "The night is dark",
    "The stars are shining",
]

# Step 1: Preprocessing (Vocabulary, Tokenization, and Padding)

# Special tokens
PAD_TOKEN = "<PAD>"
SOS_TOKEN = "<SOS>"
EOS_TOKEN = "<EOS>"

# Build vocabulary
tokens = set()
for sentence in corpus:
    tokens.update(sentence.lower().split())

vocab = [PAD_TOKEN, SOS_TOKEN, EOS_TOKEN] + sorted(list(tokens))
word2idx = {w: idx for idx, w in enumerate(vocab)}
idx2word = {idx: w for idx, w in enumerate(vocab)}

PAD_IDX = word2idx[PAD_TOKEN]
SOS_IDX = word2idx[SOS_TOKEN]
EOS_IDX = word2idx[EOS_TOKEN]
VOCAB_SIZE = len(vocab)

# Tokenize with <SOS> and <EOS>
tokenized_corpus = []
for s in corpus:
    seq = [SOS_IDX] + [word2idx[w] for w in s.lower().split()] + [EOS_IDX]
    tokenized_corpus.append(seq)

MAX_LEN = max(len(s) for s in tokenized_corpus)

# Pad sequences
padded_corpus = []
for seq in tokenized_corpus:
    padded = seq + [PAD_IDX] * (MAX_LEN - len(seq))
    padded_corpus.append(padded)

data_tensor = torch.tensor(padded_corpus, dtype=torch.long)
print(f"Vocabulary Size: {VOCAB_SIZE}")
print(f"Padded Input Tensor Shape: {data_tensor.shape}\n")



# Step 2: Build the Simple Text VAE Model

class TextVAE(nn.Module):

    def __init__(self, vocab_size, embed_dim, hidden_dim, latent_dim):
        super(TextVAE, self).__init__()
        self.embedding = nn.Embedding(
            vocab_size, embed_dim, padding_idx=PAD_IDX
        )

        # Encoder: compresses input tokens into hidden representations
        self.encoder_gru = nn.GRU(embed_dim, hidden_dim, batch_first=True)
        self.fc_mu = nn.Linear(hidden_dim, latent_dim)
        self.fc_logvar = nn.Linear(hidden_dim, latent_dim)

        # Decoder: maps latent sample back to token sequence
        self.latent_to_hidden = nn.Linear(latent_dim, hidden_dim)
        self.decoder_gru = nn.GRU(embed_dim, hidden_dim, batch_first=True)
        self.fc_out = nn.Linear(hidden_dim, vocab_size)

    def reparameterize(self, mu, logvar):
        """Reparameterization trick to allow backpropagation through random sampling."""
        std = torch.exp(0.5 * logvar)
        eps = torch.randn_like(std)
        return mu + eps * std

    def forward(self, x, teacher_forcing_inputs):
        # 1. Encode
        x_emb = self.embedding(x)
        _, h_n = self.encoder_gru(x_emb)  # h_n shape: (1, Batch, hidden_dim)
        h_last = h_n.squeeze(0)

        # 2. Get Gaussian parameters and sample latent vector z
        mu = self.fc_mu(h_last)
        logvar = self.fc_logvar(h_last)
        z = self.reparameterize(mu, logvar)

        # 3. Decode
        dec_init_hidden = torch.tanh(self.latent_to_hidden(z)).unsqueeze(
            0
        )  # (1, Batch, hidden_dim)
        dec_inputs_emb = self.embedding(teacher_forcing_inputs)
        dec_out, _ = self.decoder_gru(dec_inputs_emb, dec_init_hidden)
        logits = self.fc_out(dec_out)  # Shape: (Batch, Seq_Len, vocab_size)

        return logits, mu, logvar

    def reconstruct_greedy(self, x, max_len=6):
        """Greedy decoding using the latent mean mu."""
        self.eval()
        with torch.no_grad():
            x_emb = self.embedding(x)
            _, h_n = self.encoder_gru(x_emb)
            mu = self.fc_mu(h_n.squeeze(0))

            # Initialize decoder with the mean vector
            hidden = torch.tanh(self.latent_to_hidden(mu)).unsqueeze(0)
            curr_token = torch.tensor([[SOS_IDX]])
            generated = []

            for _ in range(max_len):
                emb = self.embedding(curr_token)
                out, hidden = self.decoder_gru(emb, hidden)
                next_token = self.fc_out(out).argmax(dim=-1).item()
                if next_token == EOS_IDX:
                    break
                generated.append(idx2word[next_token])
                curr_token = torch.tensor([[next_token]])

            return " ".join(generated)

    def generate_from_latent(self, z, max_len=6):
        """Generate a sequence by sampling an arbitrary latent vector z."""
        self.eval()
        with torch.no_grad():
            hidden = torch.tanh(self.latent_to_hidden(z)).unsqueeze(0)
            curr_token = torch.tensor([[SOS_IDX]])
            generated = []

            for _ in range(max_len):
                emb = self.embedding(curr_token)
                out, hidden = self.decoder_gru(emb, hidden)
                next_token = self.fc_out(out).argmax(dim=-1).item()
                if next_token == EOS_IDX:
                    break
                generated.append(idx2word[next_token])
                curr_token = torch.tensor([[next_token]])

            return " ".join(generated)


# Step 3: Train the Model

EMBED_DIM = 32
HIDDEN_DIM = 64
LATENT_DIM = 16

model = TextVAE(VOCAB_SIZE, EMBED_DIM, HIDDEN_DIM, LATENT_DIM)
optimizer = optim.Adam(model.parameters(), lr=0.01)
criterion = nn.CrossEntropyLoss(ignore_index=PAD_IDX)

# Decoder inputs (<SOS> ... w_{T-1}) and targets (w_1 ... <EOS>)
dec_inputs = data_tensor[:, :-1]
targets = data_tensor[:, 1:]

print("--- Training Progress ---")
model.train()
for epoch in range(1, 351):
    optimizer.zero_grad()

    logits, mu, logvar = model(data_tensor, dec_inputs)

    # 1. Reconstruction Loss (Cross Entropy)
    recon_loss = criterion(
        logits.reshape(-1, VOCAB_SIZE), targets.reshape(-1)
    )

    # 2. KL Divergence: 0.5 * sum(1 + log(sigma^2) - mu^2 - sigma^2)
    kl_div = -0.5 * torch.sum(1 + logvar - mu.pow(2) - logvar.exp())
    kl_div = kl_div / data_tensor.size(0)

    # Small KL weight (beta) to encourage reconstruction on tiny data
    total_loss = recon_loss + 0.01 * kl_div
    total_loss.backward()
    optimizer.step()

    if epoch % 50 == 0:
        print(
            f"Epoch {epoch:3d} | Total Loss: {total_loss.item():.4f} | Recon Loss: {recon_loss.item():.4f} | KL: {kl_div.item():.4f}"
        )


# Step 4: Reconstruction and Generation Evaluation

print("\n--- 1. Reconstruction of Original Sentences ---")
for i, sentence in enumerate(corpus):
    inp = data_tensor[i : i + 1]
    recon = model.reconstruct_greedy(inp)
    print(f"Original:     '{sentence}'")
    print(f"Reconstructed: '{recon}'")
    print("-" * 40)

print("\n--- 2. Generation by Sampling from Prior N(0, I) ---")
for i in range(4):
    z_sample = torch.randn(1, LATENT_DIM)
    gen = model.generate_from_latent(z_sample)
    print(f"Sample {i+1}: '{gen}'")
```

**2. Explanation**

Data Preprocessing & Padding:

The raw sentences are converted to lower-case and split into individual word tokens.

Three special tokens are added: <PAD> (index 0) for uniform sequence length, <SOS> (Start-of-Sequence) to signal decoding initiation, and <EOS> (End-of-Sequence) to mark sequence completion.

Sentences are converted to integer index vectors and padded to match the maximum sequence length.

Model Architecture:

Encoder: An embedding layer followed by a single-layer GRU. The final hidden state compresses the input sequence.

Latent Space: Two linear layers map the hidden state into two parameter vectors: $\boldsymbol{\mu}$ (mean) and $\log(\boldsymbol{\sigma}^2)$ (log-variance).

Reparameterization Trick: The latent representation $\mathbf{z}$ is sampled using $\mathbf{z} = \boldsymbol{\mu} + \boldsymbol{\sigma} \odot \boldsymbol{\epsilon}$, where $\boldsymbol{\epsilon} \sim \mathcal{N}(0, \mathbf{I})$, ensuring operations remain differentiable.

Decoder: Projects $\mathbf{z}$ into the initial hidden state of a GRU. Using teacher forcing during training, the decoder predicts the next word at each time step.

Loss Function Formulation:

The model minimizes the Evidence Lower Bound (ELBO):$$\mathcal{L} = \mathcal{L}_{\text{Reconstruction}} + \beta \cdot D_{\text{KL}}(q_\phi(\mathbf{z}\mid\mathbf{x}) \parallel \mathcal{N}(0, \mathbf{I}))$$

Reconstruction loss measures cross-entropy between predicted word logits and the ground truth tokens (ignoring <PAD>).

KL Divergence penalizes deviation from a standard Gaussian prior. A small weight ($\beta = 0.01$) is used to balance reconstruction accuracy on small datasets.

```python
'''
**Question 7: Use a pre-trained GPT model (like GPT-2 or GPT-3) to translate a short English paragraph into French and German. Provide the original and translated text.**

'''
import torch
from transformers import GPT2LMHeadModel, GPT2Tokenizer

# Load pre-trained GPT-2 model and tokenizer
model_name = "gpt2-medium"  # gpt2, gpt2-medium, or gpt2-large
tokenizer = GPT2Tokenizer.from_pretrained(model_name)
model = GPT2LMHeadModel.from_pretrained(model_name)

# Set padding token to EOS token
tokenizer.pad_token = tokenizer.eos_token
model.eval()


def translate_with_gpt2(prompt_text, target_language, stop_word="\n"):
    # Few-shot exemplar framing to guide the causal language model
    few_shot_context = (
        "Translate the English text into French and German accurately.\n\n"
        "English: Good morning, welcome to our store.\n"
        "French: Bonjour, bienvenue dans notre magasin.\n"
        "German: Guten Morgen, willkommen in unserem Geschäft.\n\n"
        "English: Technology is advancing rapidly every year.\n"
        "French: La technologie progresse rapidement chaque année.\n"
        "German: Die Technologie schreitet jedes Jahr schnell voran.\n\n"
        f"English: {prompt_text}\n"
        f"{target_language}:"
    )

    inputs = tokenizer(few_shot_context, return_tensors="pt")

    with torch.no_grad():
        output = model.generate(
            **inputs,
            max_new_tokens=40,
            temperature=0.2,
            top_p=0.9,
            repetition_penalty=1.2,
            do_sample=True,
            pad_token_id=tokenizer.eos_token_id,
        )

    # Decode and extract target translation
    generated_text = tokenizer.decode(
        output[0][inputs["input_ids"].shape[1] :], skip_special_tokens=True
    )
    # Extract only the immediate translated line
    translation = generated_text.strip().split("\n")[0]
    return translation


# Source paragraph
source_paragraph = (
    "Artificial intelligence is transforming industries across the globe. "
    "It helps businesses make smarter decisions and improves daily life."
)

french_translation = translate_with_gpt2(source_paragraph, "French")
german_translation = translate_with_gpt2(source_paragraph, "German")

print("--- Translation Results ---")
print(f"Original English:\n{source_paragraph}\n")
print(f"Translated to French:\n{french_translation}\n")
print(f"Translated to German:\n{german_translation}")
```

```python
'''
Question 8: Implement a simple attention-based encoder-decoder model for
English-to-Spanish translation using Tensorflow or PyTorch.
'''

import torch
import torch.nn as nn
import torch.nn.functional as F

# 1. Toy Dataset & Simple Vocabulary

pairs = [
    ("hello", "hola"),
    ("good morning", "buenos dias"),
    ("how are you", "como estas"),
    ("thank you", "gracias"),
    ("the sky is blue", "el cielo es azul")
]

# Build vocabulary sets
src_vocab = {"<PAD>": 0, "<SOS>": 1, "<EOS>": 2}
tgt_vocab = {"<PAD>": 0, "<SOS>": 1, "<EOS>": 2}

for eng, esp in pairs:
    for w in eng.split():
        if w not in src_vocab:
            src_vocab[w] = len(src_vocab)
    for w in esp.split():
        if w not in tgt_vocab:
            tgt_vocab[w] = len(tgt_vocab)

# Reverse dictionary to convert IDs back to words
idx_to_spanish = {idx: word for word, idx in tgt_vocab.items()}


# 2. Attention-Based Seq2Seq Model

class SimpleNMT(nn.Module):
    def __init__(self, src_size, tgt_size, hidden_dim=32):
        super().__init__()
        self.hidden_dim = hidden_dim

        # Encoder components
        self.enc_emb = nn.Embedding(src_size, hidden_dim)
        self.enc_rnn = nn.GRU(hidden_dim, hidden_dim, batch_first=True)

        # Decoder components
        self.dec_emb = nn.Embedding(tgt_size, hidden_dim)
        self.dec_rnn = nn.GRU(hidden_dim * 2, hidden_dim, batch_first=True) # takes [token_emb + context]
        self.fc_out = nn.Linear(hidden_dim, tgt_size)

    def forward_encoder(self, src_ids):
        # src_ids: (1, seq_len)
        embedded = self.enc_emb(src_ids)
        encoder_outputs, hidden = self.enc_rnn(embedded)
        return encoder_outputs, hidden

    def decode_step(self, dec_token_id, hidden, encoder_outputs):
        # 1. Simple Dot-Product Attention
        # Calculate similarity between current decoder hidden state and all encoder tokens
        scores = torch.bmm(encoder_outputs, hidden.permute(1, 2, 0)) # (1, seq_len, 1)
        attn_weights = F.softmax(scores, dim=1)

        # 2. Weighted context vector
        context = torch.sum(attn_weights * encoder_outputs, dim=1, keepdim=True) # (1, 1, hidden_dim)

        # 3. Pass [token embedding + context] into Decoder GRU
        token_emb = self.dec_emb(dec_token_id) # (1, 1, hidden_dim)
        rnn_input = torch.cat((token_emb, context), dim=-1) # (1, 1, 2*hidden_dim)

        output, hidden = self.dec_rnn(rnn_input, hidden)
        logits = self.fc_out(output.squeeze(1)) # (1, tgt_size)

        return logits, hidden


# 3. Train the Model

model = SimpleNMT(len(src_vocab), len(tgt_vocab))
optimizer = torch.optim.Adam(model.parameters(), lr=0.01)
criterion = nn.CrossEntropyLoss()

print("Training model...")
for epoch in range(1, 151):
    total_loss = 0
    for eng, esp in pairs:
        optimizer.zero_grad()

        # Prepare inputs as tensors
        src_ids = torch.tensor([[src_vocab[w] for w in eng.split()] + [src_vocab["<EOS>"]]])
        tgt_ids = [tgt_vocab[w] for w in esp.split()] + [tgt_vocab["<EOS>"]]

        # 1. Run Encoder
        enc_outputs, hidden = model.forward_encoder(src_ids)

        # 2. Run Decoder step-by-step
        curr_token = torch.tensor([[tgt_vocab["<SOS>"]]])
        loss = 0
        for target_id in tgt_ids:
            logits, hidden = model.decode_step(curr_token, hidden, enc_outputs)
            target_tensor = torch.tensor([target_id])
            loss += criterion(logits, target_tensor)
            curr_token = target_tensor.unsqueeze(0) # Teacher forcing

        loss.backward()
        optimizer.step()
        total_loss += loss.item()

print("Training finished!\n")


# 4. Translate / Test Function

def translate(eng_sentence):
    model.eval()
    with torch.no_grad():
        src_ids = torch.tensor([[src_vocab[w] for w in eng_sentence.split()] + [src_vocab["<EOS>"]]])
        enc_outputs, hidden = model.forward_encoder(src_ids)

        curr_token = torch.tensor([[tgt_vocab["<SOS>"]]])
        translated_words = []

        for _ in range(8): # Maximum translation length
            logits, hidden = model.decode_step(curr_token, hidden, enc_outputs)
            best_id = logits.argmax().item()

            if best_id == tgt_vocab["<EOS>"]:
                break
            translated_words.append(idx_to_spanish[best_id])
            curr_token = torch.tensor([[best_id]])

        return " ".join(translated_words)

# Test the trained model
print("--- Translation Results ---")
for eng, _ in pairs:
    print(f"English: {eng:<20} -> Spanish: {translate(eng)}")
```

    Minimal Attention: Replaces complex linear projection layers with Dot-Product Attention (scores = torch.bmm(encoder_outputs, hidden)), which directly measures similarity without extra trainable weights.

```python
'''
Question 9: Use the following short poetry dataset to simulate poem generation with a
pre-trained GPT model:
["Roses are red, violets are blue,",
"Sugar is sweet, and so are you.",
"The moon glows bright in silent skies,",
"A bird sings where the soft wind sighs."]
Using this dataset as a reference for poetic structure and language, generate a new 2-4
line poem using a pre-trained GPT model (such as GPT-2). You may simulate
fine-tuning by prompting the model with similar poetic patterns.
'''

import torch
from transformers import GPT2LMHeadModel, GPT2Tokenizer

model_name = "gpt2"
tokenizer = GPT2Tokenizer.from_pretrained(model_name)
model = GPT2LMHeadModel.from_pretrained(model_name)
tokenizer.pad_token = tokenizer.eos_token
model.eval()

# Complete the poem couplet prompt
prompt = (
    "Poem 1:\n"
    "Roses are red, violets are blue,\n"
    "Sugar is sweet, and so are you.\n\n"
    "Poem 2:\n"
    "The moon glows bright in silent skies,\n"
    "A bird sings where the soft wind sighs.\n\n"
    "Poem 3:\n"
    "The stars shine bright above the sea,\n"
    "And"
)

inputs = tokenizer(prompt, return_tensors="pt")

# do_sample=False locks greedy search (purely deterministic, zero randomness)
with torch.no_grad():
    output = model.generate(
        **inputs,
        max_new_tokens=12,
        do_sample=False,
        pad_token_id=tokenizer.eos_token_id,
    )

# Decode full text and isolate Poem 3
full_text = tokenizer.decode(output[0], skip_special_tokens=True)
poem_lines = [
    line.strip()
    for line in full_text.split("Poem 3:\n")[-1].split("\n")
    if line.strip()
][:2]

print("--- Generated Poem ---")
for line in poem_lines:
    print(line)
```

**Question 10: Imagine you are building a creative writing assistant for a publishing company. The assistant should generate story plots and character descriptions using Generative AI. Describe how you would design the system, including model selection, training data, bias mitigation, and evaluation methods. Explain the real-world challenges you might face.**

**Answer:**

**System Architecture**

The assistant is structured as a modular, human-in-the-loop writing environment consisting of four layers:

Interactive Interface & Author Workbench: A text workspace allowing writers to request plot outlines, brainstorm character arcs, or modify specific narrative beats.

Context Engine (Story Bible & Knowledge Graph): A vector database (e.g., Qdrant or Milvus) paired with a knowledge graph to track characters, family trees, magic systems, and world-building rules, ensuring strict continuity across long manuscripts.

Orchestration & Prompt Pipeline: An execution layer (built with LangChain or custom async Python) utilizing structured JSON schemas to parse raw story concepts into concrete character sheets (goals, flaws, backstories) and plot frameworks (Three-Act Structure, Hero's Journey).

Safety & Guardrail Layer: Pre- and post-generation filters to inspect inputs and outputs for content safety, copyright leakage, and quality benchmarks.

**Model Selection**

A hybrid approach using an enterprise-grade frontier model (e.g., Claude 3.5 Sonnet or GPT-4o) via API for high-level creative planning and nuanced prose, coupled with an open-weights model (e.g., Llama 3 70B or Mistral Large) deployed on private infrastructure for cost-effective iteration and to ensure publisher IP remains confidential.

Dense retrieval models (such as text-embedding-3-large or BGE-M3) to index the ongoing manuscript, character dossiers, and reference style guides.

**Training Data**

Curate public domain literary masterworks (Project Gutenberg), award-winning licensed manuscripts, narrative theory texts (e.g., Campbell, McKee), and professional screenplay outlines. Exclude raw web crawl text to avoid low-effort prose.

Parameter-Efficient Fine-Tuning (PEFT / LoRA): Fine-tune the base model on paired datasets of premise-to-outline transformations, dialogue balancing, and character depth templates, instilling the publishing house’s editorial guidelines.

Retrieval-Augmented Generation (RAG): Rather than updating weights for every new book, dynamically feed the current "Story Bible" into the prompt context so the model generates ideas grounded in the specific book's established lore.

**Bias Mitigation**

Clean training corpora of outdated tropes, racial caricatures, and harmful demographic stereotypes historically prevalent in classic literature.

System-Level Steering: Enforce counter-stereotypical prompting directives (e.g., explicitly instructing the generator to avoid assigning emotional volatility or domestic subservience to female characters, or villainy exclusively to marginalized archetypes).

Adversarial Red-Teaming: Subject the system to targeted stress testing with prompts likely to trigger harmful tropes, tracking demographic representation distributions across generated protagonists and antagonists.

**Evaluation Methods**

Use Natural Language Inference models to test whether a newly generated plot beat logically contradicts earlier established facts in the Story Bible.

Lexical Diversity & Repetition: Track Self-BLEU, distinct n-grams, and perplexity to detect repetitive prose loops or formulaic storytelling.

LLM-as-a-Judge: Deploy a separate, strictly prompted evaluator model to grade outputs on a 1–5 scale across explicit criteria: Pacing, Character Agency, Thematic Cohesion, and Conflict Resolution.

Human Editorial Benchmarking: Blind A/B evaluation by professional acquisition editors and novelists assessing original voice, narrative hook strength, and commercial viability.

**Real-World Challenges**

LLMs struggle to maintain global continuity across hundreds of pages, frequently forgetting subplots, swapping minor character names, or violating previously established physical rules.

Cliché and Trope Collapse: Autoregressive language models predict high-probability tokens, making them inherently prone to generic, predictable ideas (e.g., "an ancient prophecy", "eyes like obsidian pools") rather than fresh, subversive narrative twists.

Copyright and Legal Exposure: A risk that the model reconstructs recognizable plot elements, character names, or verbatim phrasing from copyrighted contemporary bestsellers present in its pre-training data.

Author Agency and UX Friction: Writers often reject generative tools that feel too prescriptive. The system must act as an unobtrusive brainstorming partner that suggests modular alternatives rather than dictating the narrative direction.
